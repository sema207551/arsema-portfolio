```python
from google.colab import files
import pandas as pd
uploaded = files.upload()
df = pd.read_csv('bank_loan (3).csv')
```

```python
# look at the first 5 rows
df.head()
```

```python
# see column types and non-null counts
df.info()

df.describe(include="all")

df.isnull().sum()

df.duplicated().sum()
```

```python
# seperate numeric and categorical columns
num_cols = df.select_dtypes(include=["int64", "float64"]).columns
cat_cols = df.select_dtypes(include="object").columns

# Fill numeric missing values with median
df[num_cols] = df[num_cols].fillna(df[num_cols].median())

# Fill categorical missing values with mode
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])
```

I filled numeric missing values with the median to lessen the effect of outliers, and categorical missing values with the mode because it perserves more common category.

```python
# how many duplicate rows?
df.duplicated().sum()
```

There was 0 duplicate rows, so there was no need to remove any rows.

```python
# recompute cat_cols in case types changed
cat_cols = df.select_dtypes(include="object").columns

# Strip spaces and make lowercase
df[cat_cols] = df[cat_cols].apply(lambda x: x.str.strip().str.lower())
```

I standardized text columns by shortening spaces and changing to lowercase so that values like "Approved" and "approved" are treated the same.
